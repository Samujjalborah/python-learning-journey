## Objective

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
df = pd.read_csv("Hotel_dataset.csv")
```

Data Quality Assessment

```python
print("="*60)
print("MISSING VALUES")
print("="*60)

missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": round(df.isnull().mean()*100,2)
})

display(missing[missing["Missing Values"]>0])

print("\n")

print("="*60)
print("DUPLICATE RECORDS")
print("="*60)

print(df.duplicated().sum())

print("\n")

print("="*60)
print("DATA TYPES")
print("="*60)

display(df.dtypes)
```

Data Cleaning & Preprocessing

```python
# Create a copy
df_clean = df.copy()

# ----------------------------
# Convert date columns
# ----------------------------

df_clean["Booking_Date"] = pd.to_datetime(df_clean["Booking_Date"])
df_clean["Arrival_Date"] = pd.to_datetime(df_clean["Arrival_Date"])
df_clean["Reservation_Status_Date"] = pd.to_datetime(df_clean["Reservation_Status_Date"])

# ----------------------------
# Remove duplicate rows
# ----------------------------

print("Rows before removing duplicates:", len(df_clean))

df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))

# ----------------------------
# Missing value handling
# ----------------------------

# Numerical columns

df_clean["Children"] = df_clean["Children"].fillna(df_clean["Children"].median())

df_clean["ADR"] = df_clean["ADR"].fillna(df_clean["ADR"].median())

df_clean["Satisfaction_Score"] = df_clean["Satisfaction_Score"].fillna(
    df_clean["Satisfaction_Score"].median()
)

# Categorical columns

df_clean["Hotel_Location"] = df_clean["Hotel_Location"].fillna(
    df_clean["Hotel_Location"].mode()[0]
)

df_clean["Country"] = df_clean["Country"].fillna(
    df_clean["Country"].mode()[0]
)

df_clean["Meal_Type"] = df_clean["Meal_Type"].fillna(
    df_clean["Meal_Type"].mode()[0]
)

# Agent ID

df_clean["Agent_ID"] = df_clean["Agent_ID"].fillna(0)

# Company ID

df_clean["Company_ID"] = df_clean["Company_ID"].fillna(0)

print("\nRemaining Missing Values\n")

display(df_clean.isnull().sum())
```

```python
# ----------------------------
# Standardize categorical values
# ----------------------------

df_clean["Hotel_Type"] = (
    df_clean["Hotel_Type"]
    .str.strip()
    .str.title()
)

df_clean["Market_Segment"] = (
    df_clean["Market_Segment"]
    .str.strip()
    .str.title()
)

df_clean["Hotel_Location"] = (
    df_clean["Hotel_Location"]
    .str.strip()
    .str.title()
)

# Fix inconsistent names after title case
df_clean["Market_Segment"] = df_clean["Market_Segment"].replace({
    "Online Ta": "Online TA",
    "Offline Ta/To": "Offline TA/TO"
})

print("Hotel Types:")
print(df_clean["Hotel_Type"].unique())

print("\nMarket Segments:")
print(df_clean["Market_Segment"].unique())
```

Bivariate Analysis

Revenue by Hotel Type

```python
hotel_summary = (
    df_clean.groupby("Hotel_Type")
    .agg(
        Average_Revenue=("Estimated_Revenue","mean"),
        Average_ADR=("ADR","mean"),
        Average_Satisfaction=("Satisfaction_Score","mean")
    )
)

display(hotel_summary)

plt.figure(figsize=(6,4))
sns.barplot(data=df_clean,
            x="Hotel_Type",
            y="Estimated_Revenue")
plt.title("Average Revenue by Hotel Type")
plt.show()
```

Revenue by Market Segment

```python
segment_summary = (
    df_clean.groupby("Market_Segment")
    .agg(
        Records=("Booking_ID","count"),
        Average_Revenue=("Estimated_Revenue","mean"),
        Cancellation_Rate=("Is_Canceled","mean")
    )
    .sort_values("Average_Revenue",ascending=False)
)

display(segment_summary)

plt.figure(figsize=(10,5))
sns.barplot(
    data=df_clean,
    x="Market_Segment",
    y="Estimated_Revenue",
    estimator="mean"
)

plt.xticks(rotation=45)
plt.title("Average Revenue by Market Segment")
plt.show()
```

Revenue by Customer Type

```python
customer_summary = (
    df_clean.groupby("Customer_Type")
    .agg(
        Records=("Booking_ID","count"),
        Average_Revenue=("Estimated_Revenue","mean"),
        Average_Satisfaction=("Satisfaction_Score","mean")
    )
)

display(customer_summary)

plt.figure(figsize=(8,4))
sns.boxplot(
    data=df_clean,
    x="Customer_Type",
    y="Estimated_Revenue"
)

plt.xticks(rotation=20)
plt.title("Revenue by Customer Type")
plt.show()
```

# Executive Business Insights

1. Resort Hotels generate substantially higher revenue than City Hotels.
   - The average estimated revenue for Resort Hotels (~784.52) is more than twice that of City Hotels (~385.52), indicating that resort properties contribute significantly more revenue per booking.

2. Longer stays contribute to higher revenue.
   - Total Nights has a strong positive correlation with Estimated Revenue (r = 0.838), suggesting that encouraging longer guest stays can significantly increase hotel revenue.

3. Room pricing strongly impacts revenue.
   - ADR (Average Daily Rate) has a moderately strong positive correlation with Estimated Revenue (r = 0.603), showing that pricing strategy plays an important role in revenue generation.

4. Booking cancellations remain a major operational challenge.
   - Reservation status analysis shows that canceled bookings account for the largest proportion of reservations, indicating revenue uncertainty and operational inefficiencies.

5. Customer satisfaction is relatively consistent across customer types.
   - Average satisfaction scores remain close to 3.85 across all customer segments, suggesting a fairly uniform guest experience irrespective of customer category.

# Management Recommendations

1. Increase focus on Resort Hotels by expanding premium services and targeted marketing campaigns, as these properties generate considerably higher revenue.

2. Introduce promotional packages that encourage guests to book longer stays, since longer stays are strongly associated with increased revenue.

3. Review cancellation policies and implement strategies such as flexible pricing, booking reminders, or loyalty incentives to reduce the high cancellation rate.

4. Optimize ADR using demand-based pricing while maintaining customer satisfaction to maximize revenue potential.

5. Continue maintaining service quality across all customer segments, as satisfaction scores are consistently high.

6. Monitor booking channels and market segments regularly to identify the most profitable customer sources and allocate marketing resources efficiently.

7. Perform periodic data quality audits to standardize categorical variables and maintain accurate reporting for future business analysis.

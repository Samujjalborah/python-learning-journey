# Day 11 - Cleaned Company Employee Dataset using Pandas

## Objective

This notebook demonstrates the process of cleaning a messy company employee dataset using the Pandas library. The dataset is inspected for missing values, duplicate records, incorrect data types, and inconsistencies. Appropriate data-cleaning techniques are applied to produce a clean dataset ready for further analysis.

**Developed by:** Samujjal Borah

```python
import pandas as pd
df = pd.read_csv("Company_Employee.csv")
df.head()
```

```python
print(df.shape)
```

Dataset Information

```python
df.info()
```

Descriptive Statistics

```python
df.describe()
```

Missing Values

```python
df.isnull().sum()
```

Duplicate Records

```python
print(df.duplicated().sum())
```

Remove Duplicates

```python
df = df.drop_duplicates()
```

Fill Missing Age (Median)

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

Fill Missing Salary (Mean)

```python
df["Annual_Salary"] = df["Annual_Salary"].fillna(df["Annual_Salary"].mean())
```

Fill Missing Experience (Median)

```python
df["Experience_Years"] = df["Experience_Years"].fillna(
    df["Experience_Years"].median()
)
```

Fill Missing Performance Score (Mean)

```python
df["Performance_Score"] = df["Performance_Score"].fillna(
    df["Performance_Score"].mean()
)
```

Fill Missing Department (Mode)

```python
df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)
```

Fill Missing Gender (Mode)

```python
df["Gender"] = df["Gender"].fillna(
    df["Gender"].mode()[0]
)
```

Fill Missing City (Mode)

```python
df["City"] = df["City"].fillna(
    df["City"].mode()[0]
)
```

Fill Missing Work Mode (Forward Fill)

```python
df["Work_Mode"] = df["Work_Mode"].ffill()
```

Convert Joining Date

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
```

Verify Dataset

```python
df.info()
```

Missing Values After Cleaning

```python
df.isnull().sum()
```

Duplicate Records After Cleaning

```python
print(df.duplicated().sum())
```

Export Cleaned Dataset

```python
df.to_csv(
    "Cleaned_Company_Employee_Dataset.csv",
    index=False
)

print("Dataset exported successfully!")
```

# Cleaning Summary

The following data-cleaning steps were performed:

- Removed duplicate records using `drop_duplicates()`.
- Filled missing values in **Age** using the median.
- Filled missing values in **Annual_Salary** using the mean.
- Filled missing values in **Experience_Years** using the median.
- Filled missing values in **Performance_Score** using the mean.
- Filled missing values in **Department**, **Gender**, and **City** using the mode.
- Filled missing values in **Work_Mode** using forward fill.
- Converted the **Joining_Date** column to DateTime format.
- Verified that the cleaned dataset contains no missing values or duplicate records.

# Conclusion

The company employee dataset was successfully cleaned using Pandas. Missing values were handled using suitable imputation techniques, duplicate records were removed, and the date column was converted into DateTime format. The resulting dataset is complete, consistent, and ready for further analysis and visualization.

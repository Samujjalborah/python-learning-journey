# Day 6 - NumPy Practice

## Objective

This notebook demonstrates the basic features of NumPy for numerical data analysis, including array creation, indexing, slicing, reshaping, mathematical operations, broadcasting, boolean masking, and aggregate functions.

**Developed by:** Samujjal Borah

## 1. Creating NumPy Arrays

```python
import numpy as np
arr1 = np.array([10,20,30,40,50])

arr2 = np.array([
    [1,2,3],
    [4,5,6]
])

arr3 = np.array([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
])

print(arr1)
print(arr2)
print(arr3)
```

## 2. Dimensions and Data Types

```python
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

print(arr1.dtype)
print(arr2.dtype)
```

## 3. Indexing and Slicing

```python
print(arr1[2])

print(arr2[1,2])

print(arr1[1:4])

print(arr2[:,1])
```

## 4. Reshaping Arrays

```python
arr = np.arange(12)

print(arr)

reshaped = arr.reshape(3,4)

print(reshaped)
```

## 5. Creating Arrays using NumPy Functions

```python
print(np.zeros((2,3)))

print(np.ones((3,2)))

print(np.arange(1,11))

print(np.linspace(0,1,5))
```

## 6. Mathematical Operations

```python
a = np.array([1,2,3])

b = np.array([4,5,6])

print(a+b)

print(a-b)

print(a*b)

print(a/b)

print(a**2)
```

## 7. Vectorized Operations

```python
arr = np.array([10,20,30])

print(arr+5)

print(arr*2)

print(np.sqrt(arr))

print(np.sin(arr))
```

## 8. Boolean Masking

```python
arr = np.array([12,5,30,18,7,25])

mask = arr>15

print(mask)

print(arr[mask])
```

## 9. Broadcasting

```python
arr = np.array([
    [1,2,3],
    [4,5,6]
])

print(arr+10)
```

## 10. Aggregate Functions

```python
arr = np.array([10,20,30,40,50])

print("Sum:",np.sum(arr))

print("Mean:",np.mean(arr))

print("Minimum:",np.min(arr))

print("Maximum:",np.max(arr))

print("Median:",np.median(arr))
```

# Measures of Central Tendency

These statistics describe the central value of a dataset.

### Mean
Mean = Σx / n

### Median
Median = Middle value after sorting the observations.

### Mode
Mode = Most frequently occurring value.

```python
variables = [
    "Weekly_Study_Hours",
    "Average_Sleep_Hours",
    "Daily_Screen_Time_Hours",
    "Stress_Score",
    "Academic_Readiness_Score"
]

central_tendency = pd.DataFrame({
    "Mean": df[variables].mean(),
    "Median": df[variables].median(),
    "Mode": df[variables].mode().iloc[0]
})

print("="*60)
print("MEASURES OF CENTRAL TENDENCY")
print("="*60)

display(central_tendency)
```

- Students sleep approximately **7 hours** per day on average, with the mean and median being almost identical.
- Daily Screen Time averages **4.50 hours**, suggesting moderate daily digital device usage.
- The average Stress Score is **4.46**, indicating a moderate stress level among students.
- Academic Readiness has the highest average value (**71.77**), showing generally good preparedness across the surveyed students.

# Measures of Dispersion

### Range
Range = Maximum − Minimum

### Variance
Variance = Σ(x − Mean)² / n

### Standard Deviation
Standard Deviation = √Variance

### Interquartile Range (IQR)

IQR = Q3 − Q1

Lower Bound = Q1 − 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

```python
dispersion = pd.DataFrame({
    "Range": df[variables].max() - df[variables].min(),
    "Variance": df[variables].var(),
    "Standard Deviation": df[variables].std(),
    "Q1": df[variables].quantile(0.25),
    "Q3": df[variables].quantile(0.75)
})

dispersion["IQR"] = dispersion["Q3"] - dispersion["Q1"]

print("="*60)
print("MEASURES OF DISPERSION")
print("="*60)

display(dispersion)
```

## Interpretation of Dispersion

- Academic Readiness Score shows the greatest variability, with a standard deviation of **9.69**, indicating the widest spread of scores among students.
- Average Sleep Hours has the smallest variability (SD = **0.84**), meaning most students have similar sleeping habits.
- Weekly Study Hours exhibit moderate variation (IQR = **5.43 hours**).
- Daily Screen Time Hours also vary considerably across students (IQR = **2.20 hours**).
- Stress Score has an IQR of **2.40**, indicating moderate differences in stress levels among students.

# Outlier Detection using IQR

Any observation below

Q1 − 1.5 × IQR

or above

Q3 + 1.5 × IQR

is considered an outlier.

```python
outlier_columns = [
    "Weekly_Study_Hours",
    "Daily_Screen_Time_Hours",
    "Commute_Time_Minutes",
    "Monthly_Discretionary_Spending"
]

outlier_summary = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    outlier_summary.append({
        "Variable": col,
        "Outliers": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

print("="*60)
print("OUTLIER SUMMARY (IQR METHOD)")
print("="*60)

display(outlier_summary)
```

## Interpretation of Outlier Analysis

The IQR method was used to detect outliers in four numerical variables.

- **Weekly Study Hours** contains **8 outliers**, indicating that a few students study substantially more or less than the majority.
- **Daily Screen Time Hours** contains **18 outliers**, suggesting that some students spend exceptionally high or low amounts of time on digital devices.
- **Commute Time Minutes** contains **4 outliers**, meaning only a small number of students have unusually long or short commutes.
- **Monthly Discretionary Spending** contains **20 outliers**, the highest among the selected variables, indicating that spending habits vary considerably for a small group of students.

Mean vs Median Before & After Removing Outliers

```python
col = "Monthly_Discretionary_Spending"

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

clean_df = df[(df[col] >= lower) & (df[col] <= upper)]

comparison = pd.DataFrame({
    "Before":[
        df[col].mean(),
        df[col].median()
    ],
    "After":[
        clean_df[col].mean(),
        clean_df[col].median()
    ]
}, index=["Mean","Median"])

print("="*60)
print("BEFORE VS AFTER REMOVING OUTLIERS")
print("="*60)

display(comparison)
```

## Effect of Removing Outliers

Monthly Discretionary Spending was selected to compare statistics before and after removing outliers.

- Mean before removing outliers: **6343.78**
- Mean after removing outliers: **5973.72**
- Median before removing outliers: **5685.50**
- Median after removing outliers: **5590.50**

The mean decreased more noticeably than the median after removing outliers, indicating that a small number of extremely high spending values were increasing the average. The median changed only slightly, demonstrating that it is more robust to outliers.

# Probability

### Probability Formula

P(A) = Number of favourable outcomes / Total number of outcomes

### Conditional Probability

P(A|B) = P(A ∩ B) / P(B)

```python
# Define Events

A = df["Part_Time_Job"] == "Yes"
B = df["Stress_Score"] >= 7
C = df["Scholarship"] == "Yes"
D = df["Exercise_Days_Per_Week"] >= 3

# Probabilities

P_A = A.mean()
P_B = B.mean()
P_C = C.mean()
P_D = D.mean()

P_A_or_B = (A | B).mean()
P_A_and_B = (A & B).mean()

P_A_given_B = (A & B).sum() / B.sum()
P_B_given_A = (A & B).sum() / A.sum()

probability = pd.DataFrame({
    "Probability":[
        P_A,
        P_B,
        P_C,
        P_D,
        P_A_or_B,
        P_A_and_B,
        P_A_given_B,
        P_B_given_A
    ]
},
index=[
    "P(A)",
    "P(B)",
    "P(C)",
    "P(D)",
    "P(A or B)",
    "P(A and B)",
    "P(A|B)",
    "P(B|A)"
])

print("="*60)
print("PROBABILITY ANALYSIS")
print("="*60)

display(probability)
```

## Probability Analysis

The probabilities of the defined events are:

- P(A): Probability that a student has a part-time job = **0.253**
- P(B): Probability that a student has a Stress Score of at least 7 = **0.075**
- P(C): Probability that a student has a scholarship = **0.307**
- P(D): Probability that a student exercises at least 3 days per week = **0.593**
- P(A ∪ B): Probability that a student has a part-time job or a high stress score = **0.277**
- P(A ∩ B): Probability that a student has both a part-time job and a high stress score = **0.052**
- P(A|B): Among students with high stress, approximately **68.9%** have a part-time job.
- P(B|A): Among students with part-time jobs, approximately **20.4%** have high stress.

Mutually Exclusive Events

```python
year1 = (df["Year_of_Study"] == 1)
year4 = (df["Year_of_Study"] == 4)

mutual = (year1 & year4).sum()

print("Students in both Year 1 and Year 4:", mutual)

if mutual == 0:
    print("Conclusion: Year 1 and Year 4 are mutually exclusive events.")
else:
    print("Conclusion: They are NOT mutually exclusive.")
```

## Mutually Exclusive Events

Year of Study = 1 and Year of Study = 4 are **mutually exclusive events** because a student cannot belong to both years simultaneously. The intersection of these two events is zero.

Independence Check

```python
print("P(A and B) =", round(P_A_and_B,4))
print("P(A) × P(B) =", round(P_A * P_B,4))

if abs(P_A_and_B - (P_A * P_B)) < 0.01:
    print("\nConclusion: Events A and B appear approximately independent.")
else:
    print("\nConclusion: Events A and B do not appear independent.")
```

## Independence of Events A and B

The probability of both events occurring together is:

- P(A ∩ B) = **0.0517**

The product of their individual probabilities is:

- P(A) × P(B) = **0.0190**

Since these values are not equal, Events A (having a part-time job) and B (Stress Score ≥ 7) do **not** appear to be independent. This suggests that having a part-time job may be associated with students experiencing higher stress levels in this dataset.

#Bayes' Theorem

```python
P_not_A = 1 - P_A

P_B_given_not_A = ((~A) & B).sum() / (~A).sum()

P_A_given_B_bayes = (P_B_given_A * P_A) / P_B

print("="*60)
print("BAYES' THEOREM")
print("="*60)

print("P(B|A):", round(P_B_given_A,4))
print("P(B|not A):", round(P_B_given_not_A,4))
print("P(A):", round(P_A,4))
print("P(B):", round(P_B,4))

print("\nP(A|B) using Bayes' Theorem:", round(P_A_given_B_bayes,4))
print("Direct Conditional Probability:", round(P_A_given_B,4))
```

## Bayes' Theorem

Bayes' Theorem was used to calculate the probability that a student has a part-time job given that the student has a Stress Score of at least 7.

- P(B|A) = **0.2039**
- P(B|not A) = **0.0312**
- P(A) = **0.2533**
- P(B) = **0.0750**

Using Bayes' Theorem:

P(A|B) = **0.6889**

The directly calculated conditional probability is also:

P(A|B) = **0.6889**

Both values are identical, confirming that the Bayes' theorem calculation is correct.

#Normal Distribution

```python
mean = df["Academic_Readiness_Score"].mean()
std = df["Academic_Readiness_Score"].std()

highest = df["Academic_Readiness_Score"].max()
lowest = df["Academic_Readiness_Score"].min()

z_high = (highest - mean) / std
z_low = (lowest - mean) / std

print("="*60)
print("NORMAL DISTRIBUTION")
print("="*60)

print("Mean:", round(mean,2))
print("Standard Deviation:", round(std,2))
print()

print("Highest Score:", highest)
print("Highest Z-score:", round(z_high,2))
print()

print("Lowest Score:", lowest)
print("Lowest Z-score:", round(z_low,2))

print("\nEmpirical Rule")
print("Within ±1 SD : 68%")
print("Within ±2 SD : 95%")
print("Within ±3 SD : 99.7%")
```

## Normal Distribution Analysis

The Academic Readiness Score has:

- Mean = **71.77**
- Standard Deviation = **9.69**

The highest score (**98.0**) has a Z-score of **2.71**, meaning it lies approximately **2.7 standard deviations above the mean**, making it an exceptionally high score.

The lowest score (**41.9**) has a Z-score of **−3.08**, indicating it is more than **3 standard deviations below the mean**, making it an unusually low score.

According to the Empirical Rule for a normal distribution:

- Approximately **68%** of observations are expected within **±1 standard deviation**.
- Approximately **95%** are expected within **±2 standard deviations**.
- Approximately **99.7%** are expected within **±3 standard deviations** of the mean.

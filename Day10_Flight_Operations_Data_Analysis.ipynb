# Day 10 - Flight Operations Data Analysis using Pandas

## Objective

This notebook performs a complete data analysis of a Flight Operations dataset using the Pandas library. The analysis includes loading the dataset, understanding its structure, selecting and filtering data, sorting records, grouping and aggregating information, applying transformations, and identifying meaningful insights from the data.

**Developed by:** Samujjal Borah

```python
import pandas as pd
df = pd.read_csv("Flight_Operations.csv")
df.head()
```

```python
df.tail()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
df.columns
```

Select Important Columns

```python
df[[
    "Airline",
    "Origin",
    "Destination",
    "Passengers",
    "Flight_Status"
]].head()
```

Filter Delayed Flights

```python
df[df["Flight_Status"]=="Delayed"]
```

Filter Flights with More Than 150 Passengers

```python
df[df["Passengers"]>150]
```

Sort by Delay Minutes

```python
df.sort_values(
    by="Delay_Minutes",
    ascending=False
).head(10)
```

Flights by Airline

```python
df.groupby("Airline")["Passengers"].sum()
```

Flights by Travel Class

```python
df.groupby("Travel_Class")["Passengers"].sum()
```

Average Delay by Airline

```python
df.groupby("Airline")["Delay_Minutes"].mean()
```

Flight Status Count

```python
df["Flight_Status"].value_counts()
```

Weather-wise Flights

```python
df.groupby("Weather")["Flight_ID"].count()
```

Average Ticket Price by Airline

```python
df.groupby("Airline")["Average_Ticket_Price"].mean()
```

Passenger Satisfaction by Airline

```python
df.groupby("Airline")["Passenger_Satisfaction"].mean()
```

Booking Channel Analysis

```python
df.groupby("Booking_Channel")["Passengers"].sum()
```

Create Occupancy Percentage

```python
df["Occupancy_%"] = (
    df["Passengers"] /
    df["Seat_Capacity"]
) * 100

df.head()
```

Convert Date

```python
df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])
```

Extract Month

```python
df["Month"] = df["Flight_Date"].dt.month_name()
```

Extract Day Name

```python
df["Day"] = df["Flight_Date"].dt.day_name()

df.head()
```

Top 10 Busiest Routes

```python
df.groupby(
    ["Origin","Destination"]
)["Passengers"].sum().sort_values(
    ascending=False
).head(10)
```

Highest Passenger Satisfaction

```python
df.sort_values(
    by="Passenger_Satisfaction",
    ascending=False
).head(10)
```

Top 10 Flights with Highest Occupancy

```python
df.sort_values(
    by="Occupancy_%",
    ascending=False
).head(10)
```

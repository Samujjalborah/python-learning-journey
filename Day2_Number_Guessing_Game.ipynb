# Day 2 - Number Guessing Game

This program generates a random number and asks the user to guess it within a limited number of attempts.

## Features

- Generates a random number
- Accepts user guesses
- Provides hints (Too High / Too Low)
- Limits the number of attempts
- Displays a success or failure message

Developed by: Samujjal Borah

```python
import random

MIN_NUMBER = 1
MAX_NUMBER = 10
MAX_ATTEMPTS = 5

secret_number = random.randint(MIN_NUMBER, MAX_NUMBER)

attempts = MAX_ATTEMPTS

while attempts > 0:
  guess = int(input(f"Guess a number between {MIN_NUMBER} and {MAX_NUMBER}: "))

  if guess == secret_number:
    print("🎉 Congratulations! You guessed the correct number.")
    break

  elif guess < secret_number:
    print("📉 Too low! Try again.")

  else:
    print("📈 Too high! Try again.")

  attempts = attempts - 1
  print(f"Remaining attempts: {attempts}")
  if attempts == 0:
    print("\n😞 Game Over!")
    print(f"The correct number was {secret_number}.")
```

## What I Learned

- Importing and using the `random` module
- Generating random numbers with `random.randint()`
- Taking user input with `input()`
- Using `if`, `elif`, and `else`
- Repeating tasks with a `while` loop
- Ending a loop with `break`
- Tracking attempts using a counter

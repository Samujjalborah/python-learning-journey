# Day 9 - Processed E-commerce Dataset using Pandas

## Objective

This notebook demonstrates how to process an e-commerce dataset by combining multiple CSV files using Pandas. The datasets are merged using `merge()`, combined using `concat()`, transformed using `apply()`, and analyzed using DateTime operations. Finally, a clean processed dataset is exported as a CSV file.

**Developed by:** Samujjal Borah

Display the datasets

```python
import pandas as pd
orders = pd.read_csv("Orders.csv")
customers = pd.read_csv("Customers.csv")
products = pd.read_csv("Prodects.csv")
orders.head()
customers.head()
products.head()
```

Dataset Information

```python
print("Orders")
orders.info()

print("\nCustomers")
customers.info()

print("\nProducts")
products.info()
```

Dataset Shape

```python
print("Orders:", orders.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
```

Merge Orders + Customers

```python
orders_customers = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="inner"
)

orders_customers.head()
```

Merge with Products

```python
final_df = pd.merge(
    orders_customers,
    products,
    on="Product_ID",
    how="inner"
)

final_df.head()
```

Demonstrate concat()

```python
top5 = final_df.head()
bottom5 = final_df.tail()

combined = pd.concat([top5, bottom5])

combined
```

Create Total Sales using apply()

```python
final_df["Total_Sales"] = final_df.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

final_df.head()
```

Create Order Size using apply()

```python
def order_size(quantity):

    if quantity <= 2:
        return "Small"

    elif quantity <= 4:
        return "Medium"

    else:
        return "Large"

final_df["Order_Size"] = final_df["Quantity"].apply(order_size)

final_df.head()
```

DateTime Operations

```python
final_df["Order_Date"] = pd.to_datetime(final_df["Order_Date"])
```

Extract Month

```python
final_df["Month"] = final_df["Order_Date"].dt.month_name()
```

Extract Day

```python
final_df["Day"] = final_df["Order_Date"].dt.day
```

Extract Day of Week

```python
final_df["Day_of_Week"] = final_df["Order_Date"].dt.day_name()

final_df.head()
```

Final Processed Dataset

```python
processed_df = final_df[[
    "Order_ID",
    "Order_Date",
    "Customer_Name",
    "City",
    "Region",
    "Membership_Type",
    "Product_Name",
    "Category",
    "Brand",
    "Quantity",
    "Unit_Price",
    "Total_Sales",
    "Payment_Method",
    "Order_Status",
    "Order_Size",
    "Month",
    "Day",
    "Day_of_Week"
]]

processed_df.head()
```

Export CSV

```python
processed_df.to_csv(
    "Processed_Ecommerce_Dataset.csv",
    index=False
)

print("Processed dataset exported successfully!")
```

# Observations

- The Orders, Customers, and Products datasets were successfully combined using `merge()`.
- `concat()` was demonstrated by combining two DataFrames.
- A new column, **Total_Sales**, was created using the `apply()` function.
- Order dates were converted into DateTime format, and Month, Day, and Day of the Week were extracted.
- The processed dataset combines customer, product, and order information into a single structured DataFrame suitable for further analysis.
